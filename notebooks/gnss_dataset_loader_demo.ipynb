## 7. 可视化训练样本波形

绘制首批训练样本的径向与垂直位移波形，检查数据预处理效果。

```python
import matplotlib.pyplot as plt

plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False
radial_np = radial.squeeze(1).cpu().numpy()
vertical_np = vertical.cpu().numpy()
distance_np = distance.cpu().numpy()
magnitude_np = magnitude.cpu().numpy()
dt_np = dt.cpu().numpy()

num_plots = min(32, radial_np.shape[0])
cols = 4
rows = int(np.ceil(num_plots / cols))
fig, axes = plt.subplots(rows, cols, figsize=(cols * 3, rows * 1.5), sharex=False)
axes = np.array(axes).reshape(-1)

for i in range(num_plots):
    ax = axes[i]
    T = radial_np.shape[1]
    t_axis = np.arange(T) * float(dt_np[i])
    ax.plot(t_axis, radial_np[i], color='tab:blue', linewidth=2, label='径向')
    ax.plot(t_axis, vertical_np[i], color='tab:orange', linewidth=2, alpha=0.8, label='垂直')
    ax.set_title(f"样本{i+1} | Mw={magnitude_np[i]:.2f}, r={distance_np[i]:.0f}m", fontsize=9)
    ax.grid(True, linestyle='--', alpha=0.3)

for j in range(num_plots, rows * cols):
    axes[j].axis('off')

fig.suptitle("训练首批样本波形（径向与垂直）", fontsize=12)
fig.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```

## 8. STF（震源时间函数）统计分析

检查 STF 矩率数据的取值范围、峰值分布，以及由 STF 积分得到的震级。

```python
import torch

stf = sample_batch.get('stf', None)
stf_log = sample_batch.get('stf_log', None)
has_stf = sample_batch.get('has_stf', None)

if stf is not None and has_stf is not None:
    stf_batch = stf if stf.dim() >= 2 else stf.unsqueeze(0)
    has_stf_mask = has_stf.to(dtype=torch.bool)
    valid_count = int(has_stf_mask.sum().item())
    
    stf_valid = stf_batch[has_stf_mask] if valid_count > 0 else stf_batch[:0]
    stf_flat = stf_valid.reshape(-1)
    stf_nonzero = stf_flat[stf_flat > 0.0]
    
    if stf_nonzero.numel() > 0:
        print(f"STF 非零范围: [{stf_nonzero.min():.3e}, {stf_nonzero.max():.3e}]")
        print(f"有效样本数: {valid_count}")
        
        # 计算 M0 和 Mw
        dt_valid = dt[has_stf_mask].to(dtype=stf_batch.dtype)
        M0_seq = torch.cumsum(torch.clamp(stf_valid, min=0.0), dim=1) * dt_valid.unsqueeze(1)
        M0_peak = torch.max(M0_seq, dim=1).values
        Mw = (2.0 / 3.0) * (torch.log10(torch.clamp(M0_peak, min=1e-10)) - 9.1)
        print(f"Mw 范围: [{Mw.min():.2f}, {Mw.max():.2f}]")
else:
    print("当前批次无 STF 数据")
```
